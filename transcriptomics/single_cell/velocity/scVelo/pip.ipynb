```python
import os
import scvelo as scv
import scanpy as sc
```

```python
adata_raw = sc.read_h5ad('/lvdata/wzb/scRNA/FW2023-504/2024.10.10/EpiFib/celltype/Result.h5ad')
```

```python
loom_dic = '/home/wzb/scRNA/FW2023-504'
outdir = '/lvdata/wzb/scRNA/FW2023-504/2024.10.10/scvelo'
dirname = os.listdir(loom_dic)
dirlist = []
for i in dirname:
    dirfile = os.path.join(loom_dic,i,'velocyto',i+".loom")
    dirlist.append(dirfile)

adatas_list = []
for i in dirlist:
    data =  scv.read_loom(i)
    data.var_names_make_unique()
    adatas_list.append(data)
adata = adatas_list[0].concatenate(adatas_list[1:], index_unique=None)
adata.write_h5ad(os.path.join(outdir,'loom.h5ad'))
```

```python
adata = scv.utils.merge(adata_raw, adata)
```

```python
outdir = '/lvdata/wzb/scRNA/FW2023-504/2024.10.10/scvelo/'


scv.pp.filter_and_normalize(adata,  min_cells =3, min_cells_u=3,min_shared_cells=3, n_top_genes=2000)
scv.pp.moments(adata, n_pcs=30, n_neighbors=30)
scv.pp.neighbors(adata)
# scv.tl.recover_dynamics(adata,n_jobs=10)
scv.tl.velocity(adata,mode="stochastic")
scv.tl.velocity_graph(adata)
#可视化
scv.pl.velocity_embedding_stream(adata, basis='umap',color='cell_type',show=False,dpi=300,save=f"{outdir}scvelo_stochastic.pdf")
scv.pl.velocity_embedding_stream(adata, basis='umap',color='cell_type',show=False,dpi=300,save=f"{outdir}scvelo_stochastic.png")


scv.pp.filter_and_normalize(adata,  min_cells =3, min_cells_u=3,min_shared_cells=3, n_top_genes=2000)
scv.pp.moments(adata, n_pcs=30, n_neighbors=30)
scv.pp.neighbors(adata)
scv.tl.recover_dynamics(adata,n_jobs=10)
scv.tl.velocity(adata)
scv.tl.velocity_graph(adata)
#可视化
scv.pl.velocity_embedding_stream(adata, basis='umap',color='cell_type',show=False,dpi=300,save=f"{outdir}scvelo_dynamical.pdf")
scv.pl.velocity_embedding_stream(adata, basis='umap',color='cell_type',show=False,dpi=300,save=f"{outdir}scvelo_dynamical.png")
```
